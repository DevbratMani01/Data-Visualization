```python


target = int(input("Enter the number to search : "))

low = 0
high = 20

attempt = 0

while low <= high:

    guess = (low + high) // 2

    attempt += 1

    answer = input(f"Is your number {guess} h/l/c: ")

    if answer == "c":
        print(f"Your answer is correct. You got in attempt {attempt}")
        break
    elif answer == "l":
        low = guess + 1
    elif answer == "h":
        high = guess - 1


```
