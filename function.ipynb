```python
def person(name , age) :
    print(f"Name : {name}")
    print(f"Age : {age}")

person("Ram" , 38)

person(name = "Shyam" , age = 27)
```

```python
def add(*args):
    total = 0

    for i in args:
        total += i

    print(f"Sum = {total}")

add(10,20)
add(10,20,40)
        
```

```python
def students(**keyways):
    for key , value in keyways.items():
        print(f"{key} : {value}")

students( name = "Shyam" , age = 29)
students( name = "Ram" , ciy = "Ayodhya")
```

```python
def convert(temp , unit):
    if unit =='C':
        return (temp *9/5 ) + 32
    elif unit =='F':
        return (temp -32) + 5/9
    else:
        return None


convert(38,76)
convert(48 , 23)


```
